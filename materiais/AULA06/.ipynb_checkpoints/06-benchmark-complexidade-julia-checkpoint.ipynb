# Benchmark de complexidade — Escalonamento de Máquina Única (Julia)

> Companion do notebook **3.5 Machine Scheduling**. Aqui geramos/lemos **várias instâncias** do
> problema $1\,|\,r_j\,|\,\sum_j T_j$ (uma máquina, com *release*, minimizando o atraso total) e
> medimos o **custo computacional de cada método** ao crescer o número de tarefas $n$.

**Métodos comparados**

| Método | Tipo | Complexidade | Garante ótimo? |
| --- | --- | --- | --- |
| FIFO / EDD / SPT | heurística (ordenação) | $O(n \log n)$ | não |
| MILO *big-M* (JuMP + HiGHS) | exato | exponencial no pior caso¹ | sim (se sem limite de tempo) |

¹ $1\,|\,r_j\,|\,\sum T_j$ é **fortemente NP-difícil**; o modelo *big-M* usa $n(n-1)/2$ binárias.

O notebook é **autocontido**: traz o gerador, o leitor de JSON, os métodos, o laço de benchmark
e os gráficos. As instâncias são lidas de uma pasta (`INSTDIR`); use o conjunto que acompanha
este material ou gere novas com o gerador abaixo.

## Preâmbulo: pacotes e solver

```julia
using JuMP, HiGHS, DataFrames, Plots, JSON, CSV
using Random, Statistics, Printf
gr()

const SOLVER = HiGHS.Optimizer
```

## Gerador de instâncias

Esquema clássico para $1\,|\,r_j\,|\,\sum T_j$:

- duração $p_j \sim U\{1,\dots,p_{\max}\}$;
- liberação $r_j \sim U\{0,\dots,\lfloor \text{rel} \cdot P \rfloor\}$, com $P = \sum_j p_j$;
- prazo $\text{due}_j = r_j + p_j + U\{0,\dots,\lfloor \text{slack} \cdot P \rfloor\}$.

Quanto menores `rel_factor`/`slack_factor`, mais apertados os prazos (mais atraso, instância
mais "interessante").

```julia
function generate_instance(n; seed = 1, pmax = 10, rel_factor = 0.5,
                            slack_factor = 0.3, prefix = "J")
    rng = MersenneTwister(1_000_003 * seed + 7 * n)
    p = rand(rng, 1:pmax, n)
    P = sum(p)
    r = rand(rng, 0:round(Int, rel_factor * P), n)
    slack = rand(rng, 0:round(Int, slack_factor * P), n)
    d = r .+ p .+ slack
    DataFrame(job = ["$prefix$i" for i in 1:n], release = r, duration = p, due = d)
end

# exemplo
generate_instance(7; seed = 1)
```

### Ler e escrever instâncias em JSON (formato universal)

O formato é o mesmo usado pelas instâncias que acompanham o material, e é lido tanto por Julia
quanto por Python.

```julia
function read_instance(path)
    d = JSON.parsefile(path)
    jobs = DataFrame(
        job      = String.(d["jobs"]),
        release  = Int.(d["release"]),
        duration = Int.(d["duration"]),
        due      = Int.(d["due"]),
    )
    return jobs, String(d["name"])
end

function write_instance(dir, jobs::DataFrame; name, meta = Dict())
    isdir(dir) || mkpath(dir)
    d = Dict(
        "name" => name, "n" => nrow(jobs),
        "jobs" => jobs.job, "release" => jobs.release,
        "duration" => jobs.duration, "due" => jobs.due,
        "generator" => meta,
    )
    open(joinpath(dir, name * ".json"), "w") do io
        JSON.print(io, d, 2)
    end
    return joinpath(dir, name * ".json")
end
```

## Métodos

### Escalonamento a partir de uma sequência e regras empíricas

```julia
function schedule_jobs(jobs::DataFrame, seq)
    rel = Dict(jobs.job .=> jobs.release)
    dur = Dict(jobs.job .=> jobs.duration)
    due = Dict(jobs.job .=> jobs.due)
    sched = DataFrame(job = String[], start = Float64[], finish = Float64[], past = Float64[])
    t = 0.0
    for j in seq
        t = max(t, rel[j]); s = t; t += dur[j]; f = t
        push!(sched, (j, s, f, max(0.0, f - due[j])))
    end
    return sched
end

# objetivo (atraso total) de uma regra empírica
function heuristic_tardiness(jobs::DataFrame, col::Symbol)
    seq = sort(jobs, col; alg = MergeSort).job
    return sum(schedule_jobs(jobs, seq).past)
end
```

### Modelo exato MILO (*big-M*)

Aqui o horizonte (`maxtime`) e o `bigM` são **dimensionados pela instância** — um $M$ apertado
melhora a relaxação linear e a velocidade. Um limite de tempo (`time_limit`) impede que
instâncias grandes travem o estudo; quando ele é atingido, registramos o *status* e o *gap*.

```julia
function machine_schedule_milo(jobs::DataFrame; bigM = nothing, time_limit = 60.0)
    JOBS  = jobs.job
    rel   = Dict(jobs.job .=> jobs.release)
    dur   = Dict(jobs.job .=> jobs.duration)
    due   = Dict(jobs.job .=> jobs.due)
    PAIRS = [(i, j) for i in JOBS for j in JOBS if i < j]

    maxtime = sum(jobs.duration) + maximum(jobs.due) + maximum(jobs.release)
    M = isnothing(bigM) ? maxtime : bigM

    model = Model(SOLVER); set_silent(model)
    set_optimizer_attribute(model, "time_limit", float(time_limit))

    @variable(model, 0 <= start[JOBS]  <= maxtime)
    @variable(model, 0 <= finish[JOBS] <= maxtime)
    @variable(model, 0 <= past[JOBS]   <= maxtime)
    @variable(model, z[PAIRS], Bin)

    @constraint(model, [j in JOBS], start[j]  >= rel[j])
    @constraint(model, [j in JOBS], finish[j] == start[j] + dur[j])
    @constraint(model, [j in JOBS], past[j]   >= finish[j] - due[j])
    @constraint(model, [p in PAIRS], finish[p[1]] <= start[p[2]] + M * (1 - z[p]))
    @constraint(model, [p in PAIRS], finish[p[2]] <= start[p[1]] + M * z[p])

    @objective(model, Min, sum(past[j] for j in JOBS))
    optimize!(model)
    return model
end
```

## Leitura das instâncias

Ajuste `INSTDIR` para a pasta onde você extraiu as instâncias do material (a que contém os
arquivos `inst_*.json`).

```julia
INSTDIR = "instances"   # <-- ajuste se necessário

paths = sort(filter(f -> endswith(f, ".json"), readdir(INSTDIR; join = true)))
println("$(length(paths)) instâncias encontradas em $(INSTDIR)/")
first(paths, 5)
```

## Laço de benchmark

Para cada instância: roda as três heurísticas e o MILO exato, medindo o **tempo de parede**
(`@elapsed`) e registrando objetivo, *status* e *gap* do exato. Ajuste `TIME_LIMIT` conforme a
sua máquina.

```julia
function benchmark(paths; time_limit = 60.0)
    res = DataFrame(instance = String[], n = Int[], nbin = Int[], method = String[],
                    objective = Float64[], time_s = Float64[],
                    status = String[], gap = Float64[])
    for path in paths
        jobs, name = read_instance(path)
        n  = nrow(jobs)
        nb = n * (n - 1) ÷ 2

        # heurísticas
        for (mname, col) in (("FIFO", :release), ("EDD", :due), ("SPT", :duration))
            t = @elapsed obj = heuristic_tardiness(jobs, col)
            push!(res, (name, n, nb, mname, float(obj), t, "heuristic", 0.0))
        end

        # exato
        local model
        t = @elapsed model = machine_schedule_milo(jobs; time_limit = time_limit)
        st  = string(termination_status(model))
        obj = has_values(model) ? objective_value(model) : NaN
        gap = try relative_gap(model) catch; NaN end
        push!(res, (name, n, nb, "MILO", obj, t, st, gap))

        @info @sprintf("%-14s n=%2d  MILO obj=%6.1f  t=%7.2fs  %s",
                       name, n, obj, t, st)
    end
    return res
end

TIME_LIMIT = 60.0
results = benchmark(paths; time_limit = TIME_LIMIT)
CSV.write("benchmark_results.csv", results)
results
```

## Análise de complexidade

### Tempo × tamanho da instância

Heurísticas custam $O(n\log n)$ — praticamente constante na escala do gráfico. O método exato
**explode** com $n$ (note o eixo $y$ em escala logarítmica).

```julia
# tempo médio por n, separando exato × heurísticas
milo = filter(:method => ==("MILO"), results)
heur = filter(:method => (m -> m in ("FIFO", "EDD", "SPT")), results)

agg_milo = sort(combine(groupby(milo, :n), :time_s => mean => :t), :n)
agg_heur = sort(combine(groupby(heur, :n), :time_s => mean => :t), :n)

plt = plot(agg_milo.n, agg_milo.t;
    marker = :circle, lw = 2, label = "MILO exato",
    yscale = :log10, xlabel = "n (tarefas)", ylabel = "tempo médio (s, log)",
    title = "Custo computacional × n", legend = :topleft, size = (820, 450))
plot!(plt, agg_heur.n, max.(agg_heur.t, 1e-7);
    marker = :square, lw = 2, label = "heurísticas (FIFO/EDD/SPT)")
plt
```

### Crescimento do modelo: número de variáveis binárias

O exato tem $n(n-1)/2$ binárias — crescimento quadrático no tamanho do modelo, que se traduz em
crescimento (super)exponencial no tempo de resolução.

```julia
nb_df = sort(unique(select(results, [:n, :nbin])), :n)
bar(string.(nb_df.n), nb_df.nbin;
    legend = false, xlabel = "n (tarefas)", ylabel = "nº de binárias  n(n−1)/2",
    title = "Tamanho do modelo exato", fillcolor = :steelblue, size = (820, 400))
```

### Qualidade das heurísticas × ótimo

Para as instâncias resolvidas até a otimalidade, medimos o **excesso relativo** da melhor
heurística sobre o ótimo: $(\text{melhor heurística} - \text{ótimo}) / \text{ótimo}$. Heurísticas
são baratas, mas pagam em qualidade.

```julia
# melhor heurística por instância
best_heur = combine(groupby(heur, :instance),
                    :objective => minimum => :heur_best,
                    :n => first => :n)
opt = filter(r -> r.status == "OPTIMAL", milo)
opt = select(opt, :instance, :objective => :opt, :n)

cmp = innerjoin(best_heur, select(opt, :instance, :opt), on = :instance)
cmp.excesso = (cmp.heur_best .- cmp.opt) ./ max.(cmp.opt, 1e-9)

scatter(cmp.n, 100 .* cmp.excesso;
    xlabel = "n (tarefas)", ylabel = "excesso da melhor heurística sobre o ótimo (%)",
    title = "Preço da rapidez", legend = false, ms = 6,
    markercolor = :darkorange, size = (820, 430))
```

### Tabela-resumo por tamanho

Tempo médio do exato, fração de instâncias resolvidas na otimalidade e excesso médio das
heurísticas, por $n$.

```julia
summary = sort(combine(groupby(milo, :n),
    :time_s => mean => :milo_t_med,
    :status => (s -> mean(s .== "OPTIMAL")) => :frac_otimo,
    :gap => (g -> mean(skipmissing(replace(g, NaN => missing)))) => :gap_med,
), :n)
summary
```

## Leitura para a aula

- **Heurísticas (FIFO/EDD/SPT)**: $O(n\log n)$, tempo desprezível mesmo para $n$ grande, mas
  **sem garantia de qualidade** — o excesso sobre o ótimo cresce e varia bastante com a instância.
- **MILO exato (*big-M*)**: dá o ótimo, porém o número de binárias cresce com $n(n-1)/2$ e o tempo
  **explode**; a partir de certo $n$ o solver atinge o limite de tempo e devolve apenas uma
  solução com *gap*.
- **Mensagem central**: existe um *trade-off* qualidade × custo. Para instâncias pequenas, use o
  exato; para grandes, heurísticas (e meta-heurísticas) tornam-se indispensáveis — eventualmente
  combinadas (heurística para *warm-start* do MILO, ou *matheuristics*).
- **Próximos passos sugeridos**: (i) testar uma reformulação mais forte (p.ex. *time-indexed* ou
  *Hull* via `DisjunctiveProgramming.jl`) e comparar a relaxação; (ii) usar a melhor heurística
  como solução inicial (`set_start_value`) e medir o ganho de tempo do exato.
